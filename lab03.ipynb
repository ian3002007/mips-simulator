```python
def run_event():
    cod = f.readline()
    cod=cod.rstrip('\n')
    if(cod==''):
        return 0
    #cod[co]=cod[co].rstrip('}')
    ins.set(cod)
    tas=cod.split(' ',1)
    
    typ=tas[0]
    kid=tas[1]
    pcp()
    decte(typ,kid)
    ref()
```

```python
def ref():
    a=format(int(rf[0][1]),"b")
    a=a.zfill(32)
    b=format(int(rf[1][1]),"b")
    b=b.zfill(32)
    c=format(int(rf[2][1]),"b")
    c=c.zfill(32)
    d=format(int(rf[3][1]),"b")
    d=d.zfill(32)
    t0.set(a)
    t1.set(b)
    s0.set(c)
    s1.set(d)
def inm(im):
    global linx
    a=linx
    b=linx+1
    c=linx+2
    d=linx+3
    linx=linx+4
    aa=format(int(a), "b")
    bb=format(int(b), "b")
    cc=format(int(c), "b")
    dd=format(int(d), "b")
    aa=aa.zfill(32)
    bb=bb.zfill(32)
    cc=cc.zfill(32)
    dd=dd.zfill(32)
    m01.set('MEM['+aa+'] = '+im[0:8])
    m02.set('MEM['+bb+'] = '+im[8:16])
    m03.set('MEM['+cc+'] = '+im[16:24])
    m04.set('MEM['+dd+'] = '+im[24:])
```

```python
def pcp():
    global lin
    lin=lin+4
    textl=format(int(lin), "b")
    textl=textl.zfill(32)
    pc.set(textl)
def decte(t,k):
        match t:
            case 'add'|'addu':
                Add(t,k)
            case 'addi'|'addiu':
                Addi(t,k)
            case 'sub':
                Sub(k)
            case 'sll':
                Sll(k)
            case 'sra'|'srl':
                Sra(t,k)
            case 'and'|'or':
                And(t,k)
            case 'nor':
                Nor(k)
            case 'lw'|'sw':
                Lw(t,k)
            case 'beq'|'bne':
                Beq(t,k)
            case 'slt'|'sltu':
                Slt(t,k)
            case 'slti':
                Slti(k)
            case 'j':
                Ju(k)
            case 'jal':
                Jal(k)
            case 'jr':
                Jr(k)
```

```python
def Add(t,k):
    global rf
    c=k.split(',')
    for i in range(len(c)):
        c[i]=c[i].lstrip(' ')
    for i in range(0,4):
        if(c[0]==rf[i][0]):
            a=i
            break
    for i in range(0,4):
        if(c[1]==rf[i][0]):
            b=i
            break
    for i in range(0,4):
        if(c[2]==rf[i][0]):
            cc=i
            break
    rf[a][1]= rf[b][1]+rf[cc][1]
    match t:
        case 'add':
            im='000000'+rf[b][2]+ rf[cc][2]+rf[a][2]+'00000100000'
        case 'addu':
            im='000000'+rf[b][2]+ rf[cc][2]+rf[a][2]+'00000100001'
    inm(im)
def Addi(t,k):
    global rf
    c=k.split(',')
    for i in range(len(c)):
        c[i]=c[i].lstrip(' ')
    for i in range(0,4):
        if(c[0]==rf[i][0]):
            a=i
            break
    for i in range(0,4):
        if(c[1]==rf[i][0]):
            b=i
            break
    rf[a][1]= rf[b][1]+int(c[2])
    bc=format(int(c[2]),"b")
    bc=bc.zfill(16)
    im='001000'+rf[b][2]+rf[a][2]+bc
    inm(im)
def  Sub(k):
    global rf
    c=k.split(',')
    for i in range(len(c)):
        c[i]=c[i].lstrip(' ')
    for i in range(0,4):
        if(c[0]==rf[i][0]):
            a=i
            break
    for i in range(0,4):
        if(c[1]==rf[i][0]):
            b=i
            break
    for i in range(0,4):
        if(c[2]==rf[i][0]):
            cc=i
            break
    rf[a][1]= rf[b][1]-rf[cc][1]
    im='000000'+rf[b][2]+ rf[cc][2]+rf[a][2]+'00000100010'
    inm(im)
def And(t,k):
    global rf
    c=k.split(',')
    for i in range(len(c)):
        c[i]=c[i].lstrip(' ')
    for i in range(0,4):
        if(c[0]==rf[i][0]):
            a=i
            break
    for i in range(0,4):
        if(c[1]==rf[i][0]):
            b=i
            break
    for i in range(0,4):
        if(c[2]==rf[i][0]):
            cc=i
            break
    match t:
        case 'and':
            rf[a][1]=rf[b][1]&rf[cc][1]
            im='000000'+rf[b][2]+ rf[cc][2]+rf[a][2]+'00000100100'
        case 'or':
            rf[a][1]=rf[b][1]|rf[cc][1]
            im='000000'+rf[b][2]+ rf[cc][2]+rf[a][2]+'00000100101'
    inm(im)
def Nor(k):
    global rf
    c=k.split(',')
    for i in range(len(c)):
        c[i]=c[i].lstrip(' ')
    for i in range(0,4):
        if(c[0]==rf[i][0]):
            a=i
            break
    for i in range(0,4):
        if(c[1]==rf[i][0]):
            b=i
            break
    for i in range(0,4):
        if(c[2]==rf[i][0]):
            cc=i
            break
    rf[a][1]=rf[b][1]|rf[cc][1]
    rf[a][1]= ~rf[a][1]+2
    im='000000'+rf[b][2]+ rf[cc][2]+rf[a][2]+'00000100111'
    inm(im)
def Slt(t,k):
    global rf
    c=k.split(',')
    for i in range(len(c)):
        c[i]=c[i].lstrip(' ')
    for i in range(0,4):
        if(c[0]==rf[i][0]):
            a=i
            break
    for i in range(0,4):
        if(c[1]==rf[i][0]):
            b=i
            break
    for i in range(0,4):
        if(c[2]==rf[i][0]):
            cc=i
            break
    if(rf[b][1]<rf[cc][1]):
        rf[a][1]=1
    else:
        rf[a][1]=0
    im='000000'+rf[b][2]+ rf[cc][2]+rf[a][2]+'00000101010'
    inm(im)
def Slti(k):
    global rf
    c=k.split(',')
    for i in range(len(c)):
        c[i]=c[i].lstrip(' ')
    for i in range(0,4):
        if(c[0]==rf[i][0]):
            a=i
            break
    for i in range(0,4):
        if(c[1]==rf[i][0]):
            b=i
            break
    if(rf[b][1]<int(c[2])):
        rf[a][1]=1
    else:
        rf[a][1]=0
    bc=format(int(c[2]),"b")
    bc=bc.zfill(16)
    im='001010'+rf[b][2]+rf[a][2]+bc
    inm(im)
def Sll(k):
    global rf
    c=k.split(',')
    for i in range(len(c)):
        c[i]=c[i].lstrip(' ')
    for i in range(0,4):
        if(c[0]==rf[i][0]):
            a=i
            break
    for i in range(0,4):
        if(c[1]==rf[i][0]):
            b=i
            break
    rf[a][1]=rf[b][1]<<int(c[2])
def Sra(t,k):
    global rf
    c=k.split(',')
    for i in range(len(c)):
        c[i]=c[i].lstrip(' ')
    for i in range(0,4):
        if(c[0]==rf[i][0]):
            a=i
            break
    for i in range(0,4):
        if(c[1]==rf[i][0]):
            b=i
            break
    rf[a][1]=rf[b][1]>>int(c[2])
def Lw(t,k):
    global rf,mem
    c=k.split(',')
    for i in range(len(c)):
        c[i]=c[i].lstrip(' ')
    d=c[1].split('(')
    d[1]=d[1].rstrip(')')
    for i in range(0,4):
        if(c[0]==rf[i][0]):
            a=i
            break
    for i in range(0,4):
        if(d[1]==rf[i][0]):
            b=i
            break
    match t:
        case 'sw':
            mem[rf[b][3]+int(d[0])-4]=rf[a][1]
            ima=rf[a][1]
            ima=format(int(ima), "b")
            ima=ima.zfill(32)
            lo=rf[b][3]+int(d[0])-4
            lo1=lo+1
            lo2=lo+2
            lo3=lo+3
            lo=format(int(lo), "b")
            lo=lo.zfill(32)
            lo1=format(int(lo1), "b")
            lo1=lo1.zfill(32)
            lo2=format(int(lo2), "b")
            lo2=lo2.zfill(32)
            lo3=format(int(lo3), "b")
            lo3=lo3.zfill(32)
            d1=format(int(d[0]), "b")
            d1=d1.zfill(16)
            me1.set('MEM['+lo+'] = '+ima[0:8])
            me2.set('MEM['+lo1+'] = '+ima[8:16])
            me3.set('MEM['+lo2+'] = '+ima[16:24])
            me4.set('MEM['+lo3+'] = '+ima[24:32])
            im='101011'+rf[b][2]+ rf[a][2]+d1
        case 'lw':
            rf[a][1]=mem[rf[b][3]+int(d[0])-4]
            d1=format(int(d[0]), "b")
            d1=d1.zfill(16)
            im='100011'+rf[b][2]+ rf[a][2]+d1
    inm(im)

            
```
